# Zomato Restaurant Data Analysis

## Zomato Restaurant Data Analysis

**Description**

This notebook performs Exploratory Data Analysis (EDA) on the Zomato Restaurant Dataset using Pandas, NumPy, Matplotlib, and Seaborn. The objective is to clean the dataset, analyze restaurant information, visualize important patterns, and derive business insights.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set(style="whitegrid")
```

## Load the Dataset

**Description**

Load the Zomato dataset into a Pandas DataFrame and display the first few records.

```
df = pd.read_csv("test_zomato.csv")

if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

df.head()
```

## Explore the Dataset

**Description**

Inspect the dataset by checking its dimensions, column names, data types, and general information.

```
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```
df.describe()
```

```
df.describe(include="all")
```

```
df = df.drop_duplicates()
```

```
df
```

## Missing Values and Duplicate Records

**Description**

Identify missing values and duplicate records before proceeding with the analysis.

```
print("Missing Values")
display(df.isnull().sum())

print("Duplicate Records")
print(df.duplicated().sum())
```

## Statistical Summary

**Description**

Generate descriptive statistics to understand the numerical features of the dataset.

```
df.describe()
```

## Unique Cities and Popular Cuisine

**Description**

Identify the number of cities, the city with the highest number of restaurants, and the most frequently occurring cuisine.

```
print("Unique Cities")
print(df["City"].nunique())

print("\nCity with Highest Restaurants")
display(df["City"].value_counts().head(1))

print("\nMost Common Cuisine")
display(df["Cuisines"].value_counts().head(1))
```

## Restaurant Rating Analysis

**Description**

Calculate the average restaurant rating and identify the restaurant with the highest number of customer votes.

```
print("Average Rating")
print(df["Aggregate rating"].mean())

print("\nRestaurant with Highest Votes")
display(df.loc[[df["Votes"].idxmax()]])
```

## Cost Analysis

**Description**

Calculate the average cost for two people across all restaurants.

```
print("Average Cost for Two")
df['Average Cost for two'] = pd.to_numeric(df['Average Cost for two'], errors='coerce')
print(df["Average Cost for two"].mean())
```

## Online Delivery Analysis

**Description**

Compare the average ratings of restaurants that provide online delivery and those that do not.

```
online_delivery = (
    df.groupby("Has Online delivery")["Aggregate rating"]
    .mean()
)

online_delivery
```

## Table Booking Analysis

**Description**

Compare the average ratings of restaurants that provide table booking and those that do not.

```
table_booking = (
    df.groupby("Has Table booking")["Aggregate rating"]
    .mean()
)

table_booking
```

## Top 10 Cities

**Description**

Identify the ten cities with the highest number of restaurants.

```
top_cities = df["City"].value_counts().head(10)

top_cities
```

## Visualization 1 – Top 10 Cities

**Description**

Create a bar chart showing the cities with the highest restaurant count.

```
plt.figure(figsize=(12,6))

top_cities.plot(kind="bar")

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")

plt.show()
```

## Visualization 2 – Aggregate Rating Distribution

**Description**

Visualize how restaurant ratings are distributed across the dataset.

```
plt.figure(figsize=(8,5))

plt.hist(df["Aggregate rating"], bins=20)

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Aggregate Rating")
plt.ylabel("Frequency")

plt.show()
```

## Visualization 3 – Online Delivery

**Description**

Display the number of restaurants offering online delivery.

```
df["Has Online delivery"].value_counts().plot(kind="bar")

plt.title("Restaurants Offering Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")

plt.show()
```

## Visualization 4 – Top 10 Cuisines

**Description**

Display the most frequently occurring cuisines in the dataset.

```
df["Cuisines"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")

plt.show()
```

## Visualization 5 – Average Rating by Price Range

**Description**

Analyze how the average restaurant rating changes across different price ranges.

```
rating_price = (
    df.groupby("Price range")["Aggregate rating"]
    .mean()
)

rating_price.plot(marker="o")

plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")

plt.show()
```

## Business Insights

**Description**

Summarize the important findings obtained from the analysis.

```
print("City with Highest Restaurant Presence:")
print(df["City"].value_counts().idxmax())

print("\nMost Popular Cuisine:")
print(df["Cuisines"].value_counts().idxmax())

print("\nAverage Ratings by Online Delivery")
display(online_delivery)

print("\nAverage Ratings by Table Booking")
display(table_booking)

print("\nAverage Restaurant Rating")
print(df["Aggregate rating"].mean())

print("\nAverage Cost for Two")
print(df["Average Cost for two"].mean())
```
